```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from braid_analysis import braid_analysis_plots
import figurefirst as fifi
```

```python
from align_course_direction_analysis import unifying_algo_analysis as uaa
from align_course_direction_analysis import unifying_algo_plots as uap
```

```python
df_trajec = pd.read_parquet('MatchedOrco_acv_trim.parquet')
```

```python
unifying_algo_data = pd.read_parquet('unifying_algo_results_MatchedOrco_acv_trim.parquet')
```

```python
unifying_algo_data
```

```python
unifying_algo_data['abs_rho'] = np.abs(unifying_algo_data['rho'])
```

```python
result = uaa.get_aligned_course_array(  unifying_algo_data,
                                        df_trajec,
                                        apply_alignment_shift=True,
                                        normalize_slope_sign=True,
                                        xlim_start=-500,
                                        xlim_end=500,
                                        sort_using='rmse_affine',
                                        return_linear_and_affine_fits=False,
                                        shift_ix_percent=1, # if less than 1, will apply only a fraction of the shift, for build animations
                                       )

ix_master, aligned_time_since_flash_arr, aligned_course_arr, aligned_roi_arr = result
```

```python
dt = np.median(np.diff(df_trajec[df_trajec.obj_id_unique_event==df_trajec.obj_id_unique_event.unique()[0]].timestamps_sec.values))
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

uap.plot_aligned_course_from_array(ix_master, 
                                   aligned_time_since_flash_arr, 
                                   aligned_course_arr, 
                                   aligned_roi_arr,
                                   aligned_flash_arr=None,
                                   aligned_linear_fits=None, 
                                   aligned_affine_fits=None,
                                   xlim_start=-500,
                                   xlim_end=500,
                                   ax=ax,
                                   clean_spines=True,
                                   course_marker_size=10,
                                   linear_marker_size=0.5,
                                   affine_marker_size=0.5,
                                   )

xticks = np.arange(-400, 500, 200)
ax.set_xticks(xticks)
ax.set_xticklabels( (xticks*dt).astype('int'))
ax.set_xlabel('Aligned time, sec')

#ax.set_yticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
#ax.set_yticklabels(['$-\pi$', '$-\pi/2$','$0$','$\pi/2$','$\pi$',])
ax.set_ylabel('Course')

fifi.mpl_functions.adjust_spines(ax, ['left', 'bottom'])
```

```python
unifying_algo_data['abs_slope'] = np.abs(unifying_algo_data.slope)
unifying_algo_data['abs_rho'] = np.abs(unifying_algo_data.rho)
```

```python
weighted_slopes = []
weighted_rho = []
weighted_axis_ratio = []
for objid in unifying_algo_data.obj_id_unique_event.unique():
    unifying_algo_data_trajec = unifying_algo_data[unifying_algo_data.obj_id_unique_event==objid]
    rho = uaa.get_weighted_value(unifying_algo_data_trajec, col='abs_rho', use='rmse_affine')
    axis_ratio = uaa.get_weighted_value(unifying_algo_data_trajec, col='axis_ratio', use='rmse_affine')
    slope = uaa.get_weighted_value(unifying_algo_data_trajec, col='abs_slope', use='rmse_affine')
    weighted_slopes.append(slope/dt)
    weighted_rho.append(rho)
    weighted_axis_ratio.append(axis_ratio)
```
