```python
import pandas as pd
import matplotlib.pyplot as plt
import figurefirst as fifi
```

```python
df = pd.read_hdf('flies_splitflow_topon_c1xwt_preprocessed_optotrigger_trimmed.hdf')
```

```python
df
```

```python
fig = plt.figure( figsize=(10,10) )
for i in range(64, 64+64):
    ax = fig.add_subplot(8,8,i+1-64)

    objid = df.obj_id_unique_event.unique()[i]
    trajec = df[df.obj_id_unique_event==objid]
    ax.scatter(trajec.x,
                trajec.y, 
                c=trajec.lights_on, cmap='bwr',
                s=1)
    
    ax.set_aspect('equal')
    fifi.mpl_functions.adjust_spines(ax, [])
    ax.set_title(objid)
    fifi.mpl_functions.set_fontsize(ax, 5)
    #20220927_170437_3462_123
```
