```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
import math

def latlon_to_cartesian(lat, lon, origin_lat, origin_lon):
    """
    Convert latitude and longitude to Cartesian coordinates (x, y) in meters
    relative to an origin point.
    
    Uses the Haversine-based approach for local tangent plane approximation.
    This is accurate for distances up to a few hundred kilometers from the origin.
    
    Parameters:
    -----------
    lat : float
        Latitude of the point in decimal degrees
    lon : float
        Longitude of the point in decimal degrees
    origin_lat : float
        Latitude of the origin point in decimal degrees
    origin_lon : float
        Longitude of the origin point in decimal degrees
    
    Returns:
    --------
    tuple : (x, y)
        x: East-West distance in meters (positive = east of origin)
        y: North-South distance in meters (positive = north of origin)
    
    Example:
    --------
    >>> # Origin at San Francisco
    >>> origin_lat, origin_lon = 37.7749, -122.4194
    >>> # Point slightly north and east
    >>> point_lat, point_lon = 37.7849, -122.4094
    >>> x, y = latlon_to_cartesian(point_lat, point_lon, origin_lat, origin_lon)
    >>> print(f"x: {x:.2f} m, y: {y:.2f} m")
    """
    # Earth's radius in meters
    R = 6371000
    
    # Convert degrees to radians
    lat_rad = math.radians(lat)
    lon_rad = math.radians(lon)
    origin_lat_rad = math.radians(origin_lat)
    origin_lon_rad = math.radians(origin_lon)
    
    # Calculate differences
    dlat = lat_rad - origin_lat_rad
    dlon = lon_rad - origin_lon_rad
    
    # Calculate x (East-West) and y (North-South) distances
    # For small distances, we can use the local tangent plane approximation
    x = R * dlon * math.cos(origin_lat_rad)  # East-West distance
    y = R * dlat  # North-South distance
    
    return x, y


def haversine_distance(lat1, lon1, lat2, lon2):
    """
    Calculate the great-circle distance between two points on Earth
    using the Haversine formula. Useful for verifying the Cartesian approximation.
    
    Parameters:
    -----------
    lat1, lon1 : float
        Latitude and longitude of first point in decimal degrees
    lat2, lon2 : float
        Latitude and longitude of second point in decimal degrees
    
    Returns:
    --------
    float : Distance in meters
    """
    R = 6371000  # Earth's radius in meters
    
    # Convert to radians
    lat1_rad = math.radians(lat1)
    lat2_rad = math.radians(lat2)
    dlat = math.radians(lat2 - lat1)
    dlon = math.radians(lon2 - lon1)
    
    # Haversine formula
    a = math.sin(dlat/2)**2 + math.cos(lat1_rad) * math.cos(lat2_rad) * math.sin(dlon/2)**2
    c = 2 * math.asin(math.sqrt(a))
    distance = R * c
    
    return distance

def wrap_angle(a):
    return np.arctan2(np.sin(a), np.cos(a))

```

```python
df_casting = pd.read_csv('albatross_circling_trajec.csv')
wind_direction = 300*np.pi/180.
```

```python
origin = df_casting.iloc[0]
```

```python
x_pos = np.array([0])
y_pos = np.array([0])

for i in range(1, df_casting.shape[0]):
    x, y = latlon_to_cartesian(df_casting.iloc[i].latitude, 
                               df_casting.iloc[i].longitude, 
                               df_casting.iloc[0].latitude, 
                               df_casting.iloc[0].longitude)

    x_pos = np.hstack((x_pos, x))
    y_pos = np.hstack((y_pos, y))

df_casting_xy = pd.DataFrame({'x': x_pos, 'y': y_pos})
```

```python
timestamps = np.cumsum(np.ones_like(df_casting_xy.x)*10) - 10
df_casting_xy['timestamps_sec'] = timestamps
```

```python
plt.plot(df_casting_xy.x.values, df_casting_xy.y.values)
```

```python
from scipy.interpolate import UnivariateSpline
```

```python
interp_timestamps_sec = np.arange(0, df_casting_xy.timestamps_sec.max(), 1)
```

```python
spline_x = UnivariateSpline(df_casting_xy.timestamps_sec, df_casting_xy.x, s=1, k=1)  # s=smoothing amount
interp_x = spline_x(interp_timestamps_sec)

spline_y = UnivariateSpline(df_casting_xy.timestamps_sec, df_casting_xy.y, s=1, k=1)  # s=smoothing amount
interp_y = spline_y(interp_timestamps_sec)
```

```python
df_casting_xy_interp = pd.DataFrame({'timestamps_sec': interp_timestamps_sec,
                                     'x': interp_x,
                                     'y': interp_y})
```

```python
plt.plot(df_casting_xy.x.values, df_casting_xy.y.values)
plt.plot(df_casting_xy_interp.x.values, df_casting_xy_interp.y.values)

tix = 80
plt.plot(df_casting_xy_interp.x.values[tix], df_casting_xy_interp.y.values[tix], 'o')

print(df_casting_xy_interp.timestamps_sec.values[tix])
```

```python
import pynumdiff
```

```python
dt = np.median(np.diff(df_casting_xy_interp.timestamps_sec.values))
_, xvel = pynumdiff.savgoldiff(df_casting_xy_interp.x, dt, [2, 3, 3])
_, yvel = pynumdiff.savgoldiff(df_casting_xy_interp.y, dt, [2, 3, 3])
```

```python
from braid_analysis import flymath

df_casting_xy_interp['xvel'] = xvel
df_casting_xy_interp['yvel'] = yvel
df_casting_xy_interp['time_relative_to_flash'] = df_casting_xy_interp['timestamps_sec'] - df_casting_xy_interp.timestamps_sec.values[tix]
df_casting_xy_interp['obj_id_unique_event'] = 'albatross_circling'
df_casting_xy_interp['obj_id_unique'] = 'albatross_circling'
df_casting_xy_interp = flymath.assign_course_and_ang_vel_to_dataframe(df_casting_xy_interp)

df_casting_xy_interp.course_smoothish -= wind_direction + np.pi
df_casting_xy_interp.course_smoothish = wrap_angle(df_casting_xy_interp.course_smoothish)
```

```python
plt.plot(df_casting_xy_interp.course_smoothish, '.', markersize=1)

ix = df_casting_xy_interp.index.values
slope = 0.11
intercept=-0.9
line = slope*ix + intercept

plt.plot(ix, wrap_angle(line), '.')

np.mean(flymath.subtract_angles(line[300:400], df_casting_xy_interp.course_smoothish.values[300:400])**2)**0.5
```
