# Demo

## Generate test data

```python
t = np.arange(0.01, 5, 0.01)
y = 5*t
course = np.sign(np.sin(y))*1.5 #+ np.pi# + np.random.normal(0,1,len(y))
course += np.random.normal(0,0.01, len(course))
plt.plot(course, '.')
```

## Do fit

```python
min_ix = 200
max_ix = len(course)-1
abs_min_ix = 120

ix = np.sort(random_integers(min_ix, max_ix, np.min([50, abs_min_ix])))
slope, intercept, AM = miop_and_affine_fit(course, ix, min_ix, max_ix, abs_min_ix)
rotation, major_axis, minor_axis, Vt, translation = decompose_affine(AM)
```

```python
plot_miop_affine_fit(course, slope, intercept, AM, min_ix)
```

# With real unsteady data -- bottom fans on

```python
bottom_on_df = pd.read_hdf('/media/caveman/Vorchard1/jaleesa_unsteady/fly_data/horizontal_split_bottom_on/c1xwt_merged_preprocessed_optotrigger_trimmed_timesincelights.hdf')

df_flash_bottom_on = bottom_on_df[bottom_on_df.intensity>0]
df_sham_bottom_on = bottom_on_df[bottom_on_df.intensity==0]

#

flash_low_bottom_on = df_flash_bottom_on.groupby("obj_id_unique_event").filter(lambda d: d.z.max()<0.25)
flash_low_bottom_on = flash_low_bottom_on[~flash_low_bottom_on.xvel.isna()]

flash_up_bottom_on  = df_flash_bottom_on.groupby("obj_id_unique_event").filter(lambda d: d.z.min()>0.25)
flash_up_bottom_on = flash_up_bottom_on[~flash_up_bottom_on.xvel.isna()]

#

sham_low_bottom_on = df_sham_bottom_on.groupby("obj_id_unique_event").filter(lambda d: d.z.max()<0.25)
sham_low_bottom_on = sham_low_bottom_on[~sham_low_bottom_on.xvel.isna()]

sham_up_bottom_on  = df_sham_bottom_on.groupby("obj_id_unique_event").filter(lambda d: d.z.min()>0.25)
sham_up_bottom_on = sham_up_bottom_on[~sham_up_bottom_on.xvel.isna()]
```

```python
# get a single trajectory
braid_df = flash_up_bottom_on
obj_id = braid_df[obj_id_key].unique()[36] # << 36 is a good demo, 40, 30, 43
trajec = braid_df[braid_df[obj_id_key]==obj_id]
trajec = trajec.dropna()
```
