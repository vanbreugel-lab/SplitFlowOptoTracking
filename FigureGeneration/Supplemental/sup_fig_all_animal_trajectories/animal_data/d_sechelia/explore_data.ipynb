```python
import pandas as pd
import matplotlib.pyplot as plt
import figurefirst as fifi
```

```python
df = pd.read_csv('actual_data/dsec07_noni.csv')
```

```python
df
```

```python
fig = plt.figure( figsize=(10,10) )
Set = 0*64
for i in range(Set, 64+Set):
    ax = fig.add_subplot(8,8,i+1-Set)

    objid = df.obj_id.unique()[i]
    trajec = df[df.obj_id==objid]
    ax.scatter(trajec.x,
                trajec.y, 
                c='black', #trajec.lights_on, cmap='bwr',
                s=1)
    
    ax.set_aspect('equal')
    fifi.mpl_functions.adjust_spines(ax, [])
    ax.set_title(objid)
    fifi.mpl_functions.set_fontsize(ax, 6)
```
