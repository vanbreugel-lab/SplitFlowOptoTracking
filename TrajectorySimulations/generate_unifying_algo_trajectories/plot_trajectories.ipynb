```python
import matplotlib.pyplot as plt
import figurefirst as fifi

import generate_unifying_trajectories as gut
```

```python
for scenario in ['laminar', 'unsteady', 'stillair']:
    fig = plt.figure(figsize=(8,8))
    
    for i in range(25):
        ax = fig.add_subplot(5,5,i+1)
    
        df_affine, df_sac = gut.generate_trajectory(scenario, 5)
        
        ax.set_aspect('equal')
        fifi.mpl_functions.adjust_spines(ax, [])
        
        #braid_analysis_plots.plot_arrowhead_trajectory(df_affine.x.values, df_affine.y.values, color='black', 
        #                                               linewidth=0.5, ax=ax, arrow_length=0.06)
        
        braid_analysis_plots.plot_arrowhead_trajectory(df_sac.x.values[0:20], df_affine.y.values[0:20], color='gray', 
                                                       linewidth=0.5, ax=ax, arrow_length=0.0)
        braid_analysis_plots.plot_arrowhead_trajectory(df_sac.x.values[20:86], df_affine.y.values[20:86], color='red', 
                                                       linewidth=0.5, ax=ax, arrow_length=0.0)
        braid_analysis_plots.plot_arrowhead_trajectory(df_sac.x.values[86:], df_affine.y.values[86:], color='black', 
                                                       linewidth=0.5, ax=ax, arrow_length=0.06)
        
    fig.savefig( 'unifying_framework_' + scenario + '.pdf' )
```
