```python
import numpy as np
import pandas as pd

import sys
sys.path.insert(1, '../../../Code/')
import WindDataProcessing

import opto_tracking
import FeatureFunctions
from datetime import date


today = date.today()
print(today)
```

# Run simulations

## Casting - Laminar (constant wind)

```python

#x = np.random.choice(12000)
x=2177
print('seed:',x)

#params 
BEHAVIOR='cast_and_surge'
TAU = 0.25
TAU_SURGE = 0.02
NOISE = 1
BIAS= 0.7
SURGE_AMP=0.42
N_TRAJS=1000
REJECT_THRESH=0
RADIUS=0
OMEGA=0

sims_df= []

sim_results = opto_tracking.run_sims(BEHAVIOR=BEHAVIOR,
    wind_condition='constant',
    SEED=x, N_TRAJS=N_TRAJS,
    DURATION=3.21,
    DT=0.01,
    TAU=TAU, NOISE=NOISE, BIAS=BIAS, 
    SURGE_AMP=SURGE_AMP,
    TAU_SURGE=TAU_SURGE,                               
    REJECT_THRESH=0,# RADIUS=0, OMEGA=0,                                 
    CONSTANT_NOISE=True,
    BOUNDS=[(-0.5, 0.5), (-0.25, 0.25), (0, 0.5)], wind_data=None,  vector_field=None, wind_positions=None, wind_data_type='CFD')


df = opto_tracking.convert_and_save_sims(sim_results, 'SimResults_tau_' + str(TAU_SURGE) + '_noise_' + str(NOISE) + '_bias_' + str(BIAS) + '_surgeamp_' + str(SURGE_AMP) +'_tau_' + str(TAU) + '_tausurge_' + str(TAU_SURGE)  + '_n_' + str(N_TRAJS) + '_seed_'+ str(x) + '_' , str(today) + '_'+ BEHAVIOR + '_constantwind' , remove_wall_hits=False, save=True)
```

```python
#calculate position normalized to point when odor turned off
df=FeatureFunctions.get_normed_pos_groupby(df)
df.to_hdf('../../../Data/Simulated_Trajectory_Data/casting_sim_constwind_n1000.hdf',key='n')
```

# Circling with CFD wind

```python
directories = ['3D_vel_1', '3D_vel_2', '3D_vel_3', '3D_vel_4', '3D_vel_5', '3D_vel_6']
prefix = '../../../Data/'
#x = np.random.choice(12000)
x=5613
print('seed:',x)
sims_df= []

#params 
BEHAVIOR='sink_and_circle'
TAU = 0.25
TAU_SURGE= 0
NOISE = 0.7
BIAS= 1.85
REJECT_THRESH = 0.3
SURGE_AMP=0
RADIUS = 0.3
OMEGA  = 6
N_TRAJS=1000


for i, directory in enumerate(directories):
    
    print('loading ~3 seconds of CFD data:')
    
    ## makes sure that any prior wind data is deleted from memory before loading new data
    try:
        del dfs
    except NameError:
        pass
    
    dfs = WindDataProcessing.load_wind_data(prefix+directory, filetype='parquet', datatype='CFD')
    
    pos = [dfs[0].x.unique(),dfs[0].y.unique(),dfs[0].z.unique()]
    
    sim_results = opto_tracking.run_sims(
        SEED=x, BEHAVIOR=BEHAVIOR, N_TRAJS=N_TRAJS,
        DURATION=3.21,
        DT=0.01,
        TAU=TAU, NOISE=NOISE, BIAS=BIAS, 
        SURGE_AMP=SURGE_AMP,
        TAU_SURGE=TAU_SURGE, REJECT_THRESH=REJECT_THRESH,
        RADIUS=RADIUS, OMEGA = OMEGA,
        CONSTANT_NOISE=True,
        wind_condition='data',
        BOUNDS=[(-0.5, 0.5), (-0.25, 0.25), (0, 0.5)], wind_data=dfs,  vector_field=dfs, wind_positions=pos, wind_data_type='CFD')

    
    df = opto_tracking.convert_and_save_sims(sim_results, '../../../Data/Simulated_Trajectory_Data/SimResults_tau_' + str(TAU_SURGE) + '_noise_' + str(NOISE) + '_bias_' + str(BIAS) + '_surgeamp_' + str(SURGE_AMP) + '_tausurge_' + str(TAU_SURGE) + '_reject_' + str(REJECT_THRESH) + '_n_' + str(N_TRAJS) + '_seed_'+ str(x) + '_' , str(today) + '_'+ BEHAVIOR + '_' + directory,remove_wall_hits=False, save=True)
    sims_df.append(df)
```

```python
alldfs = pd.concat(sims_df)
#calculate position normalized to point when odor turned off
alldfs=FeatureFunctions.get_normed_pos_groupby(alldfs)
alldfs.to_hdf('../../../Data/Simulated_Trajectory_Data/cfd_sim_results_merged_circling_'+ str(today), key='wind')
```
